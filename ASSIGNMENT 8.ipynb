```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("-" * 40)
        print(f"Name            : {self.name}")
        print(f"Employee Number : {self.number}")
        print(f"Address         : {self.address}")
        print(f"Salary          : {self.salary}")
        print(f"Post            : {self.post}")
        print(f"Year of Joining : {self.year_of_joining}")
        print(f"Education       : {self.education}")
        print(f"Experience      : {self.year_of_experience} year(s)")
        print("-" * 40)

    def update_year_of_experience(self, years=1):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to {self.year_of_experience} year(s).")

    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s salary increased by {amount}. New salary: {self.salary}")


# ---------- Single Instance Demo ----------
employee101 = Employee(
    name="Ali Khan",
    number=101,
    address="Karachi, Pakistan",
    salary=50000,
    post="Software Engineer",
    year_of_joining=2020,
    education="BS Computer Science",
    year_of_experience=4
)

# Print four attributes individually
print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

# Call all methods
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience()

# ---------- 10 Instances ----------
employees = [
    Employee("Ahmed Raza", 102, "Lahore", 45000, "Web Developer", 2021, "BSCS", 3),
    Employee("Sara Malik", 103, "Islamabad", 60000, "Project Manager", 2018, "MBA", 6),
    Employee("Bilal Ahmed", 104, "Karachi", 40000, "Data Analyst", 2022, "BS Statistics", 2),
    Employee("Hina Fatima", 105, "Multan", 55000, "HR Manager", 2019, "MBA HR", 5),
    Employee("Usman Tariq", 106, "Faisalabad", 48000, "QA Engineer", 2020, "BSSE", 4),
    Employee("Ayesha Noor", 107, "Peshawar", 70000, "Team Lead", 2017, "MSCS", 7),
    Employee("Hamza Sheikh", 108, "Quetta", 42000, "Support Engineer", 2021, "BSIT", 3),
    Employee("Zainab Iqbal", 109, "Sialkot", 65000, "Business Analyst", 2018, "MBA", 6),
    Employee("Fahad Aziz", 110, "Hyderabad", 47000, "Backend Developer", 2020, "BSCS", 4),
    Employee("Mahnoor Shah", 111, "Rawalpindi", 52000, "System Administrator", 2019, "BSIT", 5),
]

for emp in employees:
    emp.describe_employee()

# ---------- Updations ----------
for emp in employees:
    emp.increase_salary(3000)
    emp.update_year_of_experience()

# Show updated details after changes
print("\nUpdated Employee Records:")
for emp in employees:
    emp.describe_employee()
```

```python
# ---------- Functions ----------

def create_employee(name, number, address, salary, post, year_of_joining, education, year_of_experience):
    """Creates and returns an employee record as a dictionary."""
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience
    }


def describe_employee(employee):
    """Prints detailed information about an employee."""
    print("-" * 40)
    print(f"Name            : {employee['name']}")
    print(f"Employee Number : {employee['number']}")
    print(f"Address         : {employee['address']}")
    print(f"Salary          : {employee['salary']}")
    print(f"Post            : {employee['post']}")
    print(f"Year of Joining : {employee['year_of_joining']}")
    print(f"Education       : {employee['education']}")
    print(f"Experience      : {employee['year_of_experience']} year(s)")
    print("-" * 40)


def update_year_of_experience(employee, years=1):
    """Updates the employee's year of experience."""
    employee["year_of_experience"] += years
    print(f"{employee['name']}'s experience updated to {employee['year_of_experience']} year(s).")


def increase_salary(employee, amount):
    """Increases the employee's salary by a given amount."""
    employee["salary"] += amount
    print(f"{employee['name']}'s salary increased by {amount}. New salary: {employee['salary']}")


# ---------- Single Employee Demo ----------
employee101 = create_employee(
    name="Ali Khan",
    number=101,
    address="Karachi, Pakistan",
    salary=50000,
    post="Software Engineer",
    year_of_joining=2020,
    education="BS Computer Science",
    year_of_experience=4
)

# Print four attributes individually
print(employee101["name"])
print(employee101["number"])
print(employee101["address"])
print(employee101["salary"])

# Call all "methods" (functions)
describe_employee(employee101)
increase_salary(employee101, 5000)
update_year_of_experience(employee101)

# ---------- 10 Employees ----------
employees = [
    create_employee("Ahmed Raza", 102, "Lahore", 45000, "Web Developer", 2021, "BSCS", 3),
    create_employee("Sara Malik", 103, "Islamabad", 60000, "Project Manager", 2018, "MBA", 6),
    create_employee("Bilal Ahmed", 104, "Karachi", 40000, "Data Analyst", 2022, "BS Statistics", 2),
    create_employee("Hina Fatima", 105, "Multan", 55000, "HR Manager", 2019, "MBA HR", 5),
    create_employee("Usman Tariq", 106, "Faisalabad", 48000, "QA Engineer", 2020, "BSSE", 4),
    create_employee("Ayesha Noor", 107, "Peshawar", 70000, "Team Lead", 2017, "MSCS", 7),
    create_employee("Hamza Sheikh", 108, "Quetta", 42000, "Support Engineer", 2021, "BSIT", 3),
    create_employee("Zainab Iqbal", 109, "Sialkot", 65000, "Business Analyst", 2018, "MBA", 6),
    create_employee("Fahad Aziz", 110, "Hyderabad", 47000, "Backend Developer", 2020, "BSCS", 4),
    create_employee("Mahnoor Shah", 111, "Rawalpindi", 52000, "System Administrator", 2019, "BSIT", 5),
]

for emp in employees:
    describe_employee(emp)

# ---------- Updations ----------
for emp in employees:
    increase_salary(emp, 3000)
    update_year_of_experience(emp)

# Show updated details after changes
print("\nUpdated Employee Records:")
for emp in employees:
    describe_employee(emp)
```
