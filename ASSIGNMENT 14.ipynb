```python
import sqlite3
import pandas as pd

# ==========================================
# OPEN SQLITE DATABASE
# ==========================================

conn = sqlite3.connect("database.sqlite")


# ==========================================
# 1. Average TotalPayBenefits for each JobTitle
#    Exclude 'Not Provided' and 'Not provided'
# ==========================================

print("1. Average TotalPayBenefits for each JobTitle")

query1 = """
SELECT 
    JobTitle,
    AVG(TotalPayBenefits) AS AverageTotalPayBenefits
FROM Salaries
WHERE JobTitle NOT IN ('Not Provided', 'Not provided')
GROUP BY JobTitle
ORDER BY AverageTotalPayBenefits DESC;
"""

result1 = pd.read_sql(query1, conn)
display(result1)


# ==========================================
# 2. JobTitle and OvertimePay > 50000
# ==========================================

print("2. Employees with OvertimePay greater than 50000")

query2 = """
SELECT 
    JobTitle,
    OvertimePay
FROM Salaries
WHERE OvertimePay > 50000;
"""

result2 = pd.read_sql(query2, conn)
display(result2)


# ==========================================
# 3. Total number of unique employees
#    Exclude 'Not Provided' and 'Not provided'
# ==========================================

print("3. Total number of unique employees")

query3 = """
SELECT 
    COUNT(DISTINCT EmployeeName) AS UniqueEmployees
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided');
"""

result3 = pd.read_sql(query3, conn)
display(result3)


# ==========================================
# 4. Top 10 highest paid employees
# ==========================================

print("4. Top 10 highest paid employees")

query4 = """
SELECT 
    EmployeeName,
    JobTitle,
    TotalPayBenefits
FROM Salaries
ORDER BY TotalPayBenefits DESC
LIMIT 10;
"""

result4 = pd.read_sql(query4, conn)
display(result4)


# ==========================================
# 5. All employees ordered by TotalPayBenefits
#    Descending order
# ==========================================

print("5. All employees ordered by TotalPayBenefits")

query5 = """
SELECT *
FROM Salaries
ORDER BY TotalPayBenefits DESC;
"""

result5 = pd.read_sql(query5, conn)
display(result5)


# ==========================================
# CLOSE DATABASE
# ==========================================

conn.close()
```
