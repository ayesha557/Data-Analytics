```python
# ============================================================
# E-COMMERCE SALES ANALYSIS PROJECT
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Optional display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Seaborn theme
sns.set_theme(style="whitegrid")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# Change path if your file is saved somewhere else
df = pd.read_csv("ecommerce_dataset.csv")

print("=" * 70)
print("FIRST 5 ROWS OF DATASET")
print("=" * 70)

display(df.head())


# ============================================================
# 3. CHECK DATASET SIZE
# ============================================================

print("\nDataset Shape:")
print(df.shape)

print("\nTotal Rows:", df.shape[0])
print("Total Columns:", df.shape[1])


# ============================================================
# 4. CHECK DATA TYPES
# ============================================================

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(df.dtypes)


# Complete dataset information
print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


# ============================================================
# 5. DROP IRRELEVANT COLUMNS
# Latitude and Longitude are not needed for this analysis.
#
# Note:
# In the provided dataset Longitude is incorrectly written
# as 'Longtitude', so both names are included.
# ============================================================

columns_to_drop = ["Latitude", "Longitude", "Longtitude"]

df.drop(
    columns=[col for col in columns_to_drop if col in df.columns],
    inplace=True
)

print("\nColumns after dropping Latitude/Longitude:")
print(df.columns.tolist())


# ============================================================
# 6. RENAME COLUMNS
# ============================================================

df.rename(
    columns={
        "City Code": "City",
        "Order Priority Code": "Order Priority"
    },
    inplace=True
)

print("\nColumns after renaming:")
print(df.columns.tolist())


# ============================================================
# 7. CHECK DUPLICATE ROWS
# ============================================================

duplicate_count = df.duplicated().sum()

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

print("Number of duplicate rows:", duplicate_count)


# ============================================================
# 8. DROP DUPLICATES
# ============================================================

df.drop_duplicates(inplace=True)

print("Duplicates after removal:", df.duplicated().sum())
print("Dataset shape after removing duplicates:", df.shape)


# ============================================================
# 9. CHECK MISSING / NULL VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES BEFORE CLEANING")
print("=" * 70)

print(df.isnull().sum())


# ============================================================
# 10. FILL MISSING SALES/TOTAL VALUES WITH MEDIAN
# ============================================================

# According to the project requirement:
# Fill missing sales values using the median.
#
# Here 'total' represents the sales/order value.

if df["total"].isnull().sum() > 0:
    
    sales_median = df["total"].median()
    
    df["total"] = df["total"].fillna(sales_median)
    
    print("\nMissing total values filled with median:", sales_median)

else:
    print("\nNo missing values found in total column.")


# ============================================================
# 11. HANDLE OTHER MISSING VALUES
# ============================================================

# City and Order Priority contain missing values in the dataset.
# Instead of deleting those records, we will label them Unknown.

df["City"] = df["City"].fillna(-1)
df["Order Priority"] = df["Order Priority"].fillna(-1)


# ============================================================
# 12. CONVERT DATE COLUMN TO DATETIME
# ============================================================

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print("\nDate column datatype:")
print(df["date"].dtype)


# ============================================================
# 13. REPLACE CITY CODES WITH CITY NAMES
# ============================================================

city_mapping = {
    1: "Karachi",
    2: "Lahore",
    3: "Islamabad",
    4: "Faisalabad",
    5: "Quetta",
    6: "Peshawar",
    -1: "Unknown"
}

df["City"] = df["City"].map(city_mapping)

print("\nCity Names:")
print(df["City"].value_counts())


# ============================================================
# 14. REPLACE ORDER PRIORITY CODES
# ============================================================

priority_mapping = {
    0: "LOW",
    1: "MEDIUM",
    2: "HIGH",
    3: "CRITICAL",
    -1: "Unknown"
}

df["Order Priority"] = df["Order Priority"].map(priority_mapping)

print("\nOrder Priority:")
print(df["Order Priority"].value_counts())


# ============================================================
# 15. CHECK MISSING VALUES AFTER CLEANING
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES AFTER CLEANING")
print("=" * 70)

print(df.isnull().sum())


# ============================================================
# 16. DEFINE CORRECT MONTH ORDER
# ============================================================

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

df["Month"] = pd.Categorical(
    df["Month"],
    categories=month_order,
    ordered=True
)


# ============================================================
# CLEANED DATASET PREVIEW
# ============================================================

print("\n" + "=" * 70)
print("CLEANED DATASET")
print("=" * 70)

display(df.head())


# ################################################################
# EXPLORATORY DATA ANALYSIS
# ################################################################


# ============================================================
# ANALYSIS 1
# TOP 5 BEST SELLING PRODUCTS
# ============================================================

top_5_products = (
    df.groupby("product")["total"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

print("\n" + "=" * 70)
print("TOP 5 BEST SELLING PRODUCTS")
print("=" * 70)

print(top_5_products)


# Convert to dataframe for Seaborn
top_5_df = top_5_products.reset_index()

top_5_df.columns = [
    "Product",
    "Sales"
]


# -----------------------------
# BAR CHART USING SEABORN
# -----------------------------

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=top_5_df,
    x="Product",
    y="Sales"
)

plt.title(
    "Top 5 Best Selling Products",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.ticklabel_format(
    style="plain",
    axis="y"
)

plt.tight_layout()
plt.show()


# -----------------------------
# OBSERVATION
# -----------------------------

best_product = top_5_products.idxmax()
best_product_sales = top_5_products.max()

print("\nOBSERVATION:")

print(
    f"{best_product} is the highest selling product "
    f"with total sales of {best_product_sales:,.2f}."
)

print(
    "These products contribute significantly to overall revenue "
    "and should receive greater inventory and marketing attention."
)


# ============================================================
# ANALYSIS 2
# PAYMENT METHOD DISTRIBUTION
# ============================================================

payment_distribution = df["payment_method"].value_counts()

print("\n" + "=" * 70)
print("PAYMENT METHOD DISTRIBUTION")
print("=" * 70)

print(payment_distribution)


# -----------------------------
# PIE CHART USING MATPLOTLIB
# -----------------------------

plt.figure(figsize=(8, 8))

plt.pie(
    payment_distribution.values,
    labels=payment_distribution.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title(
    "Payment Method Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.axis("equal")

plt.tight_layout()
plt.show()


# -----------------------------
# OBSERVATION
# -----------------------------

top_payment = payment_distribution.idxmax()
top_payment_count = payment_distribution.max()

payment_percentage = (
    top_payment_count /
    payment_distribution.sum()
) * 100

print("\nOBSERVATION:")

print(
    f"{top_payment} is the most frequently used payment method "
    f"with {top_payment_count} orders, representing "
    f"{payment_percentage:.2f}% of all orders."
)

print(
    "The company should make sure the most frequently used payment "
    "methods remain reliable and easy for customers to use."
)


# ============================================================
# ANALYSIS 3
# MONTH-WISE SALES
# ============================================================

monthly_sales = (
    df.groupby(
        "Month",
        observed=False
    )["total"]
    .sum()
    .reindex(month_order)
)

print("\n" + "=" * 70)
print("MONTH-WISE SALES")
print("=" * 70)

print(monthly_sales)


# -----------------------------
# LINE CHART USING MATPLOTLIB
# -----------------------------

plt.figure(figsize=(13, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Month-wise Sales Trend",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(
    rotation=45
)

plt.grid(
    True,
    alpha=0.3
)

plt.ticklabel_format(
    style="plain",
    axis="y"
)

plt.tight_layout()
plt.show()


# -----------------------------
# OBSERVATION
# -----------------------------

highest_month = monthly_sales.idxmax()
highest_month_sales = monthly_sales.max()

lowest_month = monthly_sales.idxmin()
lowest_month_sales = monthly_sales.min()

print("\nOBSERVATION:")

print(
    f"{highest_month} recorded the highest sales "
    f"of {highest_month_sales:,.2f}."
)

print(
    f"{lowest_month} recorded the lowest sales "
    f"of {lowest_month_sales:,.2f}."
)

print(
    "This indicates seasonal variation in customer purchasing behavior. "
    "Marketing campaigns and inventory planning can be increased "
    "before high-sales months."
)


# ============================================================
# ANALYSIS 4
# CITY-WISE SALES
# ============================================================

city_sales = (
    df.groupby("City")["total"]
      .sum()
      .sort_values(ascending=True)
)

print("\n" + "=" * 70)
print("CITY-WISE SALES")
print("=" * 70)

print(
    city_sales.sort_values(
        ascending=False
    )
)


# Convert to dataframe for Seaborn
city_sales_df = city_sales.reset_index()

city_sales_df.columns = [
    "City",
    "Sales"
]


# -----------------------------
# HORIZONTAL BAR CHART
# USING SEABORN
# -----------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    data=city_sales_df,
    x="Sales",
    y="City"
)

plt.title(
    "City-wise Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Total Sales")
plt.ylabel("City")

plt.ticklabel_format(
    style="plain",
    axis="x"
)

plt.tight_layout()
plt.show()


# -----------------------------
# OBSERVATION
# -----------------------------

known_city_sales = city_sales.drop(
    labels="Unknown",
    errors="ignore"
)

best_city = known_city_sales.idxmax()
best_city_sales = known_city_sales.max()

lowest_city = known_city_sales.idxmin()
lowest_city_sales = known_city_sales.min()

print("\nOBSERVATION:")

print(
    f"{best_city} has the highest total sales "
    f"of {best_city_sales:,.2f}."
)

print(
    f"{lowest_city} has the lowest total sales "
    f"of {lowest_city_sales:,.2f}."
)

print(
    "High-performing cities may receive additional inventory and "
    "marketing investment, while lower-performing cities can be "
    "targeted through promotional campaigns."
)


# ============================================================
# ANALYSIS 5
# MONTHLY SALES BY CATEGORY
# ============================================================

monthly_category_sales = (
    df.groupby(
        ["Month", "category"],
        observed=False
    )["total"]
    .sum()
    .unstack(fill_value=0)
    .reindex(month_order)
)

print("\n" + "=" * 70)
print("MONTHLY SALES BY CATEGORY")
print("=" * 70)

display(monthly_category_sales)


# -----------------------------
# SIDE-BY-SIDE BAR CHART
# USING MATPLOTLIB
# -----------------------------

ax = monthly_category_sales.plot(
    kind="bar",
    figsize=(15, 7),
    width=0.8
)

plt.title(
    "Monthly Sales by Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(
    rotation=45
)

plt.legend(
    title="Category"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.ticklabel_format(
    style="plain",
    axis="y"
)

plt.tight_layout()
plt.show()


# -----------------------------
# OBSERVATION
# -----------------------------

category_total_sales = (
    df.groupby("category")["total"]
      .sum()
      .sort_values(ascending=False)
)

best_category = category_total_sales.idxmax()
best_category_sales = category_total_sales.max()

print("\nOBSERVATION:")

print(
    f"{best_category} generated the highest overall category sales "
    f"of {best_category_sales:,.2f}."
)

print(
    "Monthly category sales fluctuate across the year, showing that "
    "different categories perform better during different periods."
)

print(
    "The business can use these seasonal patterns to plan category-specific "
    "discounts, advertisements, and inventory levels."
)


# ============================================================
# FINAL SALES SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"""
1. Best Selling Product:
   {best_product}
   Sales = {best_product_sales:,.2f}

2. Most Popular Payment Method:
   {top_payment}
   Orders = {top_payment_count}

3. Highest Sales Month:
   {highest_month}
   Sales = {highest_month_sales:,.2f}

4. Lowest Sales Month:
   {lowest_month}
   Sales = {lowest_month_sales:,.2f}

5. Highest Performing City:
   {best_city}
   Sales = {best_city_sales:,.2f}

6. Highest Performing Category:
   {best_category}
   Sales = {best_category_sales:,.2f}
"""
)


# ============================================================
# OPTIONAL: SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "cleaned_ecommerce_dataset.csv",
    index=False
)

print(
    "Cleaned dataset successfully saved as "
    "'cleaned_ecommerce_dataset.csv'"
)
```
