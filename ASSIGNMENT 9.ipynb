```python
import csv

input_file = "student data.csv"
output_file = "student data with grades.csv"

def get_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

with open(input_file, "r", newline="", encoding="utf-8") as infile:
    reader = csv.DictReader(infile)

    # Add the new Grade column
    fieldnames = reader.fieldnames + ["Grade"]

    with open(output_file, "w", newline="", encoding="utf-8") as outfile:
        writer = csv.DictWriter(outfile, fieldnames=fieldnames)

        writer.writeheader()

        for row in reader:
            percentage = float(row["Percentage"])
            row["Grade"] = get_grade(percentage)
            writer.writerow(row)

print("New CSV file created:", output_file)
```
